## Conditions (if/else)

```python
##Take a number and check whether it is positive or negative.

num=int(input("enter a number"))
if num>0:
    print("positive")
elif num==0:
    print("number is zero")    
else:
    print("negative")
```

```python
##Check whether a number is even or odd.

num=int(input("enter a number"))
if num%2==0:
    print("even")
else:
    print("odd")
```

```python
##Take two numbers and print the greater number.

num1=int(input("enter a number"))
num2=int(input("enter a number"))

if num1>num2:
    print(num1,"is greater")
elif num1==num2:
    print("both are equal numbers")
else:
    print(num2,"is greater")
```

```python
##Take three numbers and find the largest.

num1=int(input("enter a number"))
num2=int(input("enter a number"))
num3=int(input("enter a number"))

if num1>=num2 and num1>=num3:
    print(num1,"is greater")
elif num2>=num1 and num2>=num3:
    print(num2,"is greater")
else:
    print(num3,"is greater")
```

```python
##Check whether a person is eligible to vote based on age.

age=int(input("enter your age"))
if age>=18:
    print("eligible to vote")
else:
    print("not eligible to vote")
```

```python
## Check whether a number is divisible by 5.

num=int(input("enter a number"))
if num%5==0:
    print(num,"is divisible by 5")
else:
    print(num,"is not divisible by 5")
```

```python
## Check whether a number is divisible by both 3 and 5.

num=int(input("enter a number"))
if num%3==0 and num%5==0:
    print(num,"is divisible by 3 and 5")
else:
    print(num,"is not divisible by 3 and 5")
```

```python
## Take marks and print:
#90+ → A
#75–89 → B
#60–74 → C
#below 60 → Fail

marks=int(input("enter marks"))
if marks>=90:
    print("A")
elif marks>=75:
    print("B")
elif marks>=60:
    print("C")
else:
    print("Fail ")
```

```python
##Check whether a year is a leap year.
#1
n=int(input("enter the year"))
if n%400==0 or n%4==0 and n%100!=0:
    print("leap year")
else:
    print("not a leap year")
```

```python
#2
year=int(input("enter year"))

if year%400==0:
    print("leap year")
elif year%4==0 and year%100!=0:
    print("leap year")
else:
    print("not a leap year")
```

## Loops

```python
##Print numbers from 1 to 10.
for i in range(1,11):
    print(i)
```

```python
##Print all even numbers from 1 to 20
for i in range(1,21):
    if i%2==0:
        print(i)
```

```python
##Print all even numbers from 1 to 20
for i in range(2,21,2):
        print(i)
```

```python
##Print the multiplication table of a given number.

num=int(input("enter a number"))
for i in range(1,11):
    print(num,"*",i,"=",i*num)
```

```python
##Find the sum of numbers from 1 to n.

num=int(input("enter a number"))
sum=0
for i in range(1,num+1):
    sum=sum+i
print(sum,end="")
```

```python
##Find the sum of all even numbers from 1 to n.

num=int(input("enter a number"))
sum=0
for i in range(1,num+1):
    if i%2==0:
        sum=sum+i
print(sum)
```

```python
##ount how many numbers from 1 to 100 are divisible by 3.
count=0
for i in range(1,101):
    if i%3==0:
        count+=1
print(count)
```
