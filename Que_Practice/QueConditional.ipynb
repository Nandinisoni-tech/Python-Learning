```python
## Positive, Negative, or Zero

num=int(input("enter a number"))
if num>0:
   print("positive")
elif num<0:
     print("negative")
elif num==0:
     print("number is zero")
```

```python
## Even or Odd

num=int(input("enter a number"))
if num%2==0:
    print("number is even")
else:
    print("number is odd")
```

```python
## Eligible to Vote

age=int(input("enter a number"))
if age>=18:
    print("eligible to vote")
else:
    print("not eligible")
```

```python
## Greater of Two Numbers

num1=int(input("enter a number"))
num2=int(input("enter a number"))

if num1>num2:
    print(num1 ,"is greater")
elif num1<num2:
     print(num2 ,"is greater")
else:
    print("numbers are equal")
```

```python
## Divisible by 5

num=int(input("enter a number"))
if num%5==0:
    print("number is Divisible by 5")
else:
    print("number is not Divisible by 5")
```

```python
# Divisible by 3 and 5

num=int(input("enter a number"))
if num%3==0:
    print("Divisible by 3 but not by 5")
elif num%5==0:
    print("Divisible by 5 but not by 3")
elif num%3==0 & num%5==0:
    print("Divisible by both 3 and 5")
```

```python
## Greatest of Three Numbers

num1=int(input("enter a number"))
num2=int(input("enter a number"))
num3=int(input("enter a number"))

if num1 > num2 and num1 > num3:
    print(num1 ,"is greater")
elif num2 > num1 and num2 > num3:
     print(num2 ,"is greater")
else:
    print(num3,"is greater")
```

```python
# Marks Grade
# Take user input: YES
# Input marks and print: A for 90+, B for 75–89, C for 50–74, and Fail below 50.

marks=int(input("enter a number"))
if marks>=90:
   print("A")
elif marks>=75:
   print("B")
elif marks>=50:
   print("C")
else:
    print("Fail")
```

```python
## input a year and check whether it is a leap year.

year=int(input("enter the year"))
if year%400==0:
    print("leap year")
elif year % 100 == 0:
    print("non leap year")    
elif year%4==0:
    print("leap year")
else:
    print("non leap year")
```
