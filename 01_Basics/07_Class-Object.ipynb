# Class and Object

```python
## Class

A class is a blueprint or template used to create objects.

## Object

An object is an instance of a class.
```

```python
# syntax to create an object

#objectname = classname()
```

```python
class Car:    ## Car → Class
    pass

car1 = Car()  ## car1 → Object / Instance of Car
car2 = Car()  ## car2 → Object / Instance of Car
```

```python
class Student:
    name1="Nandini"
    name2="Muskan"
    
s1=Student()
s1=Student()
print(s1.name1)
print(s1.name2)
```

```python
class StudentData:
    ##constructor
    def __init__(self,name,rollno,marks):
         self.name=name                   ## name same everywhere
         self.rollno=rollno
         self.marks=marks
s1=StudentData("Nandini Soni",91,20)
print(s1.name,s1.rollno,s1.marks)
s2=StudentData("Muskan Singh",92,20)
print(s2.name,s2.rollno,s2.marks)
    
```

```python
class StudentData:
    def __init__(self,name,rollno,marks):
         self.full_name=name              ## name different at self.___
         self.rollno=rollno
         self.marks_obtained=marks
s1=StudentData("Nandini Soni",91,20)
print(s1.full_name,s1.rollno,s1.marks_obtained)
s2=StudentData("Muskan Singh",92,20)
print(s2.full_name,s2.rollno,s2.marks_obtained)
```
