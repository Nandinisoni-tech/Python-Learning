#  `*args and **kwargs`
`*args and **kwargs` are special Python keywords that are used to pass the variable length of arguments to a function

```python
## *args=in Python allows a function to accept any number of positional arguments. 
## Instead of specifying a fixed number of parameters, you can pass as many values.

##allows us to pass a variable number of *non-keyword* arguments to a function.

## we can write anything instead of args , *add , *num *alia etc...
def sum(*args):
    addition=0
    for i in (args):
       addition=addition+i
    return addition
print(sum(3,4,5,6,23,67,4,3,3,5,6,6))
```

```python
# **kwargs (Keyword Arguments) is a special syntax in Python that allows a function
# to accept any number of keyword arguments (key=value pairs).Inside the function,
# these arguments are stored as a dictionary.

def person(**kwargs):
   return (kwargs)
print(person(name="nandini",age=21,city="bhopal"))
```

```python
def city(**kwargs):
    for(key,value) in kwargs.items():
        print(key,"->",value)
print(city(delhi="india"))
```

```python
##### Points to remember while using `*args and **kwargs`

- order of the arguments matter(normal -> *args -> **kwargs)## when using both together
- The words “args” and “kwargs” are only a convention, you can use any name of your choice
```

## Lambda Function
A lambda function is a small anonymous function.

A lambda function can take any number of arguments, but can only have one expression.

```python
a = lambda x:x**2
a(2)
```

```python
x= lambda a,b:a+b
x(2,7)
```

## Diff between lambda vs Normal Function

- No name
- lambda has no return value(infact,returns a function)
- lambda is written in 1 line
- not reusable

Then why use lambda functions?<br>
**They are used with Higher Order Function**


```python
# check if a string has 'a'

a = lambda s:'a' in s
a('hello')
```

```python
## odd or even

a=lambda num:'even' if num%2==0 else 'odd'
a(7)
```

## Higher Order Functions

```python
# Example

def square(x):
  return x**2

def cube(x):
  return x**3

# HOF
def transform(f,L):
  output = []
  for i in L:
    output.append(f(i))

  print(output)

L = [1,2,3,4,5]

transform(lambda x:x**3,L)
```

## Map

```python
# square the items of a list
list(map(lambda x:x**2,[1,2,3,4,5]))
```

```python
# odd/even labelling of list items
L = [1,2,3,4,5]
list(map(lambda x:'even' if x%2 == 0 else 'odd',L))
```

```python
# fetch names from a list of dict

users = [
    {
        'name':'Rahul',
        'age':45,
        'gender':'male'
    },
    {
        'name':'Nitish',
        'age':33,
        'gender':'male'
    },
    {
        'name':'Ankita',
        'age':50,
        'gender':'female'
    }
]

list(map(lambda users:users['gender'],users))
```

## Filter

```python
# numbers greater than 5
L = [3,4,5,6,7]

list(filter(lambda x:x>5,L))
```

```python
# fetch fruits starting with 'a'
fruits = ['apple','guava','cherry']

list(filter(lambda x:x.startswith('a'),fruits))
```

## Reduce

```python
# find min
functools.reduce(lambda x,y:x if x>y else y,[23,11,45,10,1])
```
