# indexing

```python
l=[1,2,3,4,5]
print(l[1:3]) ##positive indexing

print(l[-1:-3:-1]) ##negative indexing
```

# Adding Items to a List

```python
# append -- add 1 item

L = [1,2,3,4,5]
L.append(True)
print(L)
```

```python
L = [1,2,3,4,5]
L.append([6,7,8]) ## if add more than one item ,still it try to store it as one item 
print(L)
```

```python
# extend -- store multiple items

L = [1,2,3,4,5]
L.extend([6,7,8])
print(L)
```

```python
L = [1,2,3,4,5]
L.extend('delhi') ## if add one item ,still it try to store it as multiple item 
print(L)
```

```python
# insert -- if we need to add at a particular index or in the middle of items

L = [1,2,3,4,5]

L.insert(4,100)
print(L)
```

# Editing items in a List

```python
# editing with indexing

l = [1,2,3,4,5]
l.index(4)
l[3]=5000
print(l)
```

```python
# editing with slicing

l = [1,2,3,4,5,6,7,8]
l[1:5]= 400,500,600
print(l)
```

# Deleting items from a List

```python
# del
L = [1,2,3,4,5]

# indexing
del L[-1]

# slicing
del L[1:3]
print(L)
```

```python
# remove-- works directly without giving index

L = [1,2,3,4,5]

L.remove(5)

print(L)
```

```python
# pop -- works on index
L = [1,2,3,4,5]

L.pop() # if we do not mention it directly removes last element

print(L)
```

```python
# clear -- it empty the list
L = [1,2,3,4,5]

L.clear()

print(L)
```

## Operations on Lists

- Arithmetic
- Membership
- Loop

```python
# Arithmetic (+ ,*)

L1 = [1,2,3,4]
L2 = [5,6,7,8]

# Concatenation/Merge
print(L1 + L2)
```

```python
print(L1*3)
```

```python
#Membership -- element is present or not

L1 = [1,2,3,4,5]
L2 = [1,2,3,4,[5,6]]

print(5 not in L1)
print([5] in L2)
print([5,6] in L2)
```

```python
# Loops  

L1 = [1,2,3,4,5]
L2 = [1,2,3,4,[5,6]]
L3 = [[[1,2],[3,4]],[[5,6],[7,8]]]

for i in L2:
  print(i)
```

```python
for i in L3:
  print(i)
```

# List Functions

```python
# len/min/max/sorted
L = [2,1,5,7,0]

print(len(L))
print(min(L))
print(max(L))
print(sorted(L,reverse=True))
L.reverse()
print(L)
```

```python
# count 
L = [1,2,1,3,4,1,5]
L.count(5)
```

```python
# index
L = [1,2,1,3,4,1,5]
L.index(1)
```

```python
# reverse
L = [2,1,5,7,0]
# permanently reverses the list
L.reverse()
print(L)
```

```python
# sort (vs sorted)
L = [2,1,5,7,0]
print(L)
print(sorted(L)) # does not permanently changes the original list
print(L)
L.sort()    # does not permanently changes the original list
print(L)
```

## 2 ways to traverse a list

- itemwise
- indexwise

```python
# itemwise

L = [1,2,3,4]

for i in L:
  print(i)
```

```python
# indexwise

L = [1,2,3,4]

for i in range(0,len(L)):
  print(L[i])
```

# Zip

The zip() function returns a zip object, which is an iterator of tuples where the first item in each passed iterator is paired together, and then the second item in each passed iterator are paired together.

If the passed iterators have different lengths, the iterator with the least items decides the length of the new iterator.

```python
# Write a program to add items of 2 lists indexwise

L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]

list(zip(L1,L2)) 
```

```python
# Write a program to add items of 2 lists indexwise

L1 = [1,2,3,4]
L2 = [-1,-2,-3,-4]

list(zip(L1,L2)) 
[i+j for i,j in zip(L1,L2)]
```

## Disadvantages of Python Lists

- Slow
- Risky usage
- eats up more memory

```python
## Risky usage

a = [1,2,3]
b = a

print(a)
print(b)

a.append(4)
print(a)
print(b)

# lists are mutable
```

```python
## Risky usage

a = [1,2,3]
b = a.copy()

print(a)
print(b)

a.append(4)
print(a)
print(b)

# lists are mutable
```
