# Operators in Python

```python
# Arithmetic Operators
# Relational Operators
# Logical Operators
# Bitwise Operators
# Assignment Operators
# Membership Operators
```

```python
# Arithmetric Operators
print(5+6)

print(5-6)

print(5*6)

print(5/2)

print(5//2)

print(5%2)

print(5**2)
```

```python
# Relational Operators
print(4>5)

print(4<5)

print(4>=8)

print(4<=4)

print(4==4)

print(4!=4)
```

```python
# Logical Operators
print(1 and 0)

print(1 or 0)

print(not 1)
```

```python
# Bitwise Operators

# bitwise and
print(2 & 3)

# bitwise or
print(2 | 3)

# bitwise xor
print(2 ^ 3)

print(~3)

print(4 >> 2)

print(5 << 2)
```

```python
# Assignment Operators

# = 
# a = 2

a = 2

# a = a % 2
a %= 2

# a++ ++a

print(a)
```

```python
# Membership Operators

# in/not in

print('D' not in 'Delhi')

print(1 in [2,3,4,5,6])
```

```python
# Guessing game
import random
jackpot=random.randint(1,50)
guess=int(input("enter no"))

while jackpot!=guess:
   if guess>jackpot:
     print(" guess lower")
   else:
     print("guess higher") 
       
   guess=int(input("enter again"))
    
else:
    print("correct guess")
```

# Nested Loops(pattern questions)

```python
for i in range(1,5):
  for j in range(1,5):
      print(i,j)
```

```python
for i in range(10):
   for j in range(i+1):
     print("*",end="")
   print()    
 
```

```python
count=int(input("enter no"))
for i in range(count):
   for j in range(i):
     print(i,end="")
   print() 
    
 
```

```python
# Break
for i in range(1,10):
  if i==8:
      break
  print(i)
```

```python
# Continue
for i in range(1,7):
  if i == 5:
    continue
  print(i)
```
